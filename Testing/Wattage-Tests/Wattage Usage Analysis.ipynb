```python
import pandas as pd
```

```python
def pureData(dataframe):
    dataframe.columns = dataframe.columns.str.strip()
    dataframe["power.draw [W]"] = (
        dataframe["power.draw [W]"].str.replace(" W", "", regex=False).astype(float)
    )
    dataframe["utilization.gpu [%]"] = (
        dataframe["utilization.gpu [%]"].str.replace(" %", "", regex=False).astype(int)
    )
    dataframe["timestamp"] = pd.to_datetime(dataframe["timestamp"])
    dataframe = dataframe.sort_values("timestamp")
    return dataframe

def getStats(filename):
    data = pureData(pd.read_csv(filename))
    power = data["power.draw [W]"]
    avg_wattage = power.mean()
    max_wattage = power.max()

    time = data["timestamp"].diff().dt.total_seconds() / 3600
    complete_wattage = (data["power.draw [W]"].shift(1) * time).sum()
    avg_wattage_per_hour = complete_wattage / time.sum()

    training_time = time.sum() * 3600 # Hours to seconds

    print(f"Average wattage: {avg_wattage:.1f} W")
    print(f"Max wattage: {max_wattage:.1f} W")
    print(f"Time Taken: {training_time:.1f} seconds | {(training_time / 60):.3f} minutes")
    if filename == "Inference-Session.csv":
        print(f"Average Wattage: {avg_wattage_per_hour:.1f} W | {(avg_wattage_per_hour / 1000):.3f} kW")
    else:
        print(f"Complete Wattage: {complete_wattage:.1f} Wh | {(complete_wattage / 1000):.3f} kWh")

print("Training Session Stats")
getStats("Training-Session.csv")

print("\n\nInference Session Stats")
getStats("Inference-Session.csv")
```

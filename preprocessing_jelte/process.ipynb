```python
import pandas as pd
import numpy as np
import os
from itertools import combinations
# from lingua import Language, LanguageDetectorBuilder
# from rich.pretty import pprint
```

```python
all_datasets_filenames = os.listdir("./output")
all_datasets = [
    pd.read_csv(f"./output/{filename}", dtype={"text": str, "label": np.int8}) for filename in all_datasets_filenames
]

all_datasets = [
    set(dataset["text"].tolist()) for dataset in all_datasets
]
```

```python
combinations_list = list(combinations(range(len(all_datasets)), 2))

for combination in combinations_list:
    dataset1 = all_datasets[combination[0]]
    dataset2 = all_datasets[combination[1]]
    intersection = set(dataset1).intersection(set(dataset2))
    print(f"{combination[0]:<6}; {combination[1]:<6}: {len(intersection):<6}; {len(dataset1):<6}; {len(dataset2):<6}")
```

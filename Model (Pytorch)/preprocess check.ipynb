```python
import pandas as pd

# process dataset
file_path = "data_huang_devansh.csv"
df = pd.read_csv(file_path)

print("Number of columns:", len(df.columns))
print("Column names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df)
```

```python
# Remove hyperlinks (not perfect, will also remove punctuation or anything else immediately after the hyperlink)
df["Content"] = df["Content"].str.replace(r"https?://www\.[^\s]+","",regex=True)

# Remove anything with the layout "@somwthing" (usernames) or "someone@example.com" (emails)
df["Content"] = df["Content"].str.replace(r"@\S+", "", regex=True)
df["Content"] = df["Content"].str.replace(r"\S+@\S+\.\S+", "", regex=True)
df["Content"] = df["Content"].str.replace(r"\s+", " ", regex=True).str.strip()

# Replace multiple consecutive spaces + newlines + tabs + carriage returns with a single space
df["Content"] = df["Content"].str.replace(r"\s+", " ", regex=True)
```
